```python
Final_Students_withID = []
def new_addmission():
  global Final_Students_withID
  Stu_name = input("Enter a student name: ")
  try:
      Stu_age = int(input("Enter a student age: "))
      if not (Stu_age>=5 and Stu_age<=18):
         return("Entering age is not valid")
  except ValueError:
      return("Invalid input. Please enter a number.")

  Stu_class = int(input("Enter a student class from 1 to 12: "))
  try:
    Gua_mo_no = int(input("Enter a guardian mobile number: "))
    if len(str(Gua_mo_no))!= 10:
      return("Entering mobile number is not valid")
  except ValueError:
      return("Invalid input. Please enter a number.")
  student_id = 100 + len(Final_Students_withID) + 1 ## giving unique id
  Final_Students_withID.append((student_id, (Stu_name, Stu_age, Stu_class, Gua_mo_no)))
  print(Final_Students_withID)
  print("Student id,name,age,stadard,phone number")
```

```python
def view_student_detail():
  Dic_Stud = dict(Final_Students_withID)
  stu_id = int(input("Enter a student ID: "))
  print("Name,Age,Stadard,Phone number")
  return(Dic_Stud[stu_id])
```

```python
def update_student_detail():
  global Dic_Stud
  global Final_Students_withID
  stu_id = int(input("Enter a student ID: "))
  Dic_Stud = dict(Final_Students_withID)
  if stu_id not in Dic_Stud:
    return("This Student id is not available")
  else:
    print("to update phone number write 3","\n to update Class(Standard) write 2")
  Update = int(input("Enter a number: "))

  List = list(Dic_Stud[stu_id]) ## tuple to list so we can update

  if Update == 3:
    Gua_mo_no = int(input("Enter a guardian mobile number: "))
    if len(str(Gua_mo_no))!= 10:
      return("Entering mobile number is not valid")
    else:
      List[3] = Gua_mo_no ## after updating
      print("Phone number is updated")

  elif Update == 2:
    Stu_class = int(input("Enter a student class from 1 to 12: "))
    List[2] = Stu_class ## after updating
    print("Student class is updated")

  else:
    print("Invalid input")

  Dic_Stud[stu_id] = tuple(List) ## again that list to tuple and save to same in that variable
  Final_Students_withID = list(Dic_Stud.items())  # dictionary to list of tuple
```

```python
update_student_detail()
```

```python
def Cancle_addmission():
  global Final_Students_withID
  stu_id = int(input("Enter a student ID (which is given in admission form): "))
  for i in Final_Students_withID[:]:
    if stu_id in i:
      Final_Students_withID.remove(i)
      print("Student is cancled")
```

```python
Passenger = []
def Routes_Bookingticket():
  global Passenger
  print("1. Mumbai to Pune - Rs. 500","\n 2. Delhi to Jaipur - Rs. 600","\n 3. Ahemdabad to Surat - Rs. 350")
  User = str(input("Want to by ticket (yes/no): "))
  if User == "yes":
    P_name = input("Enter a passenger name: ")
    P_age = int(input("Enter a passenger age: "))
    P_Mobilenumber = int(input("Enter a passenger mobile number: "))
    Route = int(input("Enter a route number: "))
    if len(Passenger)<=39 and Route == 1:
      Assign_seat = len(Passenger) + 1
    elif len(Passenger)<=39 and Route == 2:
      Assign_seat = len(Passenger) + 1
    elif len(Passenger)<=39 and Route == 3:
      Assign_seat = len(Passenger) + 1
    else:
      print("No more seat available")

    Ticket_ID = 100 + len(Passenger) + 1
    Passenger.append((Ticket_ID,(P_name,P_age,P_Mobilenumber,Assign_seat,Route)))
    print("Your ticket is booked")
    print("Ticket ID,Name,Age,Phone number,Seat number,Route")
    print(Passenger)
  else:
     print("Thank you for visit")
```

```python
def View_Tickets():
  dict_pass = dict(Passenger)
  P_ID = int(input("Enter a passenger ID(which is in ticket): "))
  print("Name,Age,Phone number,Seat number,Route")
  print(dict_pass[P_ID])
```

```python
def Cancle_Tickets():
  P_ID = int(input("Enter a passenger ID(which is in ticket): "))
  for i in Passenger[:]:
    if P_ID in i:
      Passenger.remove(i)
      print("Your ticket is cancled")
```

```python
print("1. Show available Route and Book Tickets","\n 2. View Tickets", "\n 3. Cancle Tickets")
User_Input = int(input("Enter a number: "))
if User_Input == 1:
  Routes_Bookingticket()
elif User_Input == 2:
  View_Tickets()
else:
  Cancle_Tickets()
```
